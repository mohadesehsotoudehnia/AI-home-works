```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
import pandas as pd
# 4. داده ها را به یک DataFrame.pandas تبدیل کرده و نام هر یک از ستون ها را مشخص کنید.

iris = load_iris()


df = pd.DataFrame(iris.data, columns=iris.feature_names)


df['species'] = iris.target


X = df.drop(columns=['species'])
y = df['species']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)


train_df = pd.DataFrame(X_train, columns=iris.feature_names)
train_df['species'] = y_train.to_numpy()


test_df = pd.DataFrame(X_test, columns=iris.feature_names)
test_df['species'] = y_test.to_numpy()

train_df['dataset_type'] = 'train'
test_df['dataset_type'] = 'test'


final_df = pd.concat([train_df, test_df], ignore_index=True)


print(final_df.sample(10))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
# 4. تابع چگالی احتمال۱ ویژگی های دادگان را به تفکیک داده ی آموزش و آزمون نمایش دهید

iris = datasets.load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)


df['species'] = pd.Series(iris.target)


df['species'] = df['species'].replace({0: 'setosa', 1: 'versicolor', 2: 'virginica'})


train_data, test_data = train_test_split(df, train_size=0.8, random_state=42, stratify=df['species'])


print("Train data shape:", train_data.shape)
print("Test data shape:", test_data.shape)


features = iris.feature_names
plt.figure(figsize=(12, 10))

for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.kdeplot(train_data[feature], label='Train', fill=True, color='b', alpha=0.5)
    sns.kdeplot(test_data[feature], label='Test', fill=True, color='r', alpha=0.5)
    plt.title(f'PDF of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Density')
    plt.legend()
    plt.grid(True)

plt.tight_layout()
plt.show()
```

# گسسته سازی پیوستگی

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
# کی از ویژگی های عددی را انتخاب کرده و مقادیر عددی پیوسته را به چند کلاس گسسته تبدیل کنید.

iris = datasets.load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target


species_map = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
df['species'] = df['species'].map(species_map)


train_df, test_df = train_test_split(df, test_size=0.2, random_state=43, stratify=df['species'])


feature_to_discretize = "sepal length (cm)"


bins = [df[feature_to_discretize].min(), 5.0, 6.5, df[feature_to_discretize].max()]
labels = ["کوتاه", "متوسط", "بلند"]


df["sepal_length_category"] = pd.cut(df[feature_to_discretize], bins=bins, labels=labels, include_lowest=True)


print(df[["sepal length (cm)", "sepal_length_category"]].head())


features = iris.feature_names
plt.figure(figsize=(12, 10))

for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.kdeplot(train_df[feature], label='Train', fill=True, color='b', alpha=0.5)
    sns.kdeplot(test_df[feature], label='Test', fill=True, color='r', alpha=0.5)
    plt.title(f'PDF of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Density')
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
# ویژگی های آماری مختلف را برای کلاس Setosa توصیف کنید. (از متد ()describe. بهره بگیرید.)

iris = datasets.load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target


species_map = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
df['species'] = df['species'].map(species_map)


train_df, test_df = train_test_split(df, test_size=0.2, random_state=43, stratify=df['species'])


feature_to_discretize = "sepal length (cm)"


bins = [df[feature_to_discretize].min(), 5.0, 6.5, df[feature_to_discretize].max()]
labels = ["کوتاه", "متوسط", "بلند"]


df["sepal_length_category"] = pd.cut(df[feature_to_discretize], bins=bins, labels=labels, include_lowest=True)


setosa_stats = df[df['species'] == 'setosa'].describe()
print("Statistical Description for Setosa:\n", setosa_stats)


features = iris.feature_names
plt.figure(figsize=(12, 10))

for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.kdeplot(train_df[feature], label='Train', fill=True, color='b', alpha=0.5)
    sns.kdeplot(test_df[feature], label='Test', fill=True, color='r', alpha=0.5)
    plt.title(f'PDF of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Density')
    plt.legend()

plt.tight_layout()
plt.show()
```
