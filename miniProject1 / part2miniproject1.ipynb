# **Data and original data**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        # Gradient Descent Algorithm
        for _ in range(self.n_iterations):
            y_predicted = np.dot(X, self.weights) + self.bias
            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / n_samples) * np.sum(y_predicted - y)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def predict(self, X):
        return np.dot(X, self.weights) + self.bias

    def mean_squared_error(self, y_true, y_pred):
        return np.mean((y_true - y_pred) ** 2)

    def mean_absolute_error(self, y_true, y_pred):
        return np.mean(np.abs(y_true - y_pred))

# Load the data
file_path = "/content/mp1_lr_dataset_ai4032.xlsx"  # Specify your path
df = pd.read_excel(file_path)

# Ensure df contains features and target
X = df.iloc[:, :-1].values  # Features
y = df.iloc[:, -1].values    # Target

# Normalize the data
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)
X_normalized = (X - X_mean) / X_std

# Create and train the model
model = LinearRegression(learning_rate=0.01, n_iterations=1000)
model.fit(X_normalized, y)

# Predict
y_pred = model.predict(X_normalized)

# Calculate and display results
mse = model.mean_squared_error(y, y_pred)
mae = model.mean_absolute_error(y, y_pred)

print(f'MSE: {mse}')
print(f'MAE: {mae}')

# Plot results: Comparison of actual vs predicted values
plt.figure(figsize=(14, 6))

# Scatter plot: Actual vs Predicted
plt.subplot(1, 2, 1)
plt.scatter(y, y_pred, color='blue')
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linewidth=2)
plt.title('Actual vs Predicted')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.grid()

# Error analysis plot
plt.subplot(1, 2, 2)
plt.bar(['MSE', 'MAE'], [mse, mae], color=['blue', 'orange'])
plt.title('Error Comparison')
plt.ylabel('Error Value')
plt.grid()

plt.tight_layout()
plt.show()
```

# **Types_of_Models_in_Data_Cleaning_and_Archaeology**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Load the dataset
file_path = "/content/mp1_lr_dataset_ai4032.xlsx"    # Specify your path
df = pd.read_excel(file_path)

# Ensure df contains features and target
X = df.iloc[:, :-1].values  # Features
y = df.iloc[:, -1].values    # Target

# (Example function) Remove outliers from the data
def remove_outliers(df):
    Q1 = df.quantile(0.25)
    Q3 = df.quantile(0.75)
    IQR = Q3 - Q1
    return df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]

# Original data
df_original = df.copy()
df_cleaned = remove_outliers(df)

# Features and target for cleaned data
X_cleaned = df_cleaned.iloc[:, :-1].values
y_cleaned = df_cleaned.iloc[:, -1].values

# Train the model on original data
model_original = LinearRegression()
model_original.fit(X, y)
y_pred_original = model_original.predict(X)

# Train the model on cleaned data
model_cleaned = LinearRegression()
model_cleaned.fit(X_cleaned, y_cleaned)
y_pred_cleaned = model_cleaned.predict(X_cleaned)

# Calculate model parameters
params_original = (model_original.intercept_, model_original.coef_)
params_cleaned = (model_cleaned.intercept_, model_cleaned.coef_)

# Display parameters
print(f'Parameters for original model: Intercept = {params_original[0]}, Coefficients = {params_original[1]}')
print(f'Parameters for cleaned model: Intercept = {params_cleaned[0]}, Coefficients = {params_cleaned[1]}')

# Calculate and print errors
mse_original = mean_squared_error(y, y_pred_original)
mse_cleaned = mean_squared_error(y_cleaned, y_pred_cleaned)

mae_original = mean_absolute_error(y, y_pred_original)
mae_cleaned = mean_absolute_error(y_cleaned, y_pred_cleaned)

print(f'MSE for original model: {mse_original}, MSE for cleaned model: {mse_cleaned}')
print(f'MAE for original model: {mae_original}, MAE for cleaned model: {mae_cleaned}')

# Plot results
plt.figure(figsize=(14, 6))

# Plot original results
plt.subplot(1, 2, 1)
plt.scatter(y, y_pred_original, color='blue', alpha=0.5, label='Predicted (Original)')
plt.scatter(y, y, color='red', alpha=0.5, label='Actual')
plt.title('Model on Original Data')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.legend()
plt.grid()

# Plot cleaned results
plt.subplot(1, 2, 2)
plt.scatter(y_cleaned, y_pred_cleaned, color='green', alpha=0.5, label='Predicted (Cleaned)')
plt.scatter(y_cleaned, y_cleaned, color='red', alpha=0.5, label='Actual')
plt.title('Model on Cleaned Data')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```
